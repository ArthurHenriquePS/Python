```python
def lista():
    l = []
    for i in range(0, 101):
        l.append(i)
    return l

def dois_ponteiros(l, a):
    e, d, ln = 0, len(l) - 1, []

    while e < d:
        sa = l[e] + l[d]

        if sa == a:
            ln.append([l[e], l[d]])
            e += 1
            d -= 1

        elif sa < a:
            e += 1
        elif sa > a:
            d -= 1
    return ln if ln else "Nada"

%time dois_ponteiros(lista(), 33)
    
```

```python
def quadrado_numeros(l):
    e, d = 0, len(l) - 1
    r = [0] * len(l)
    p = len(l) - 1

    while e < d:
        qe = l[e] ** 2
        qd = l[d] ** 2

        if qe > qd:
            r[p] = qe
            e += 1
        else:
            r[p] = qd
            d -= 1
        p -= 1
    return r
l = [-4, -1, 0, 3, 10]
quadrado_numeros(l)
```

```python
def inverte_vogais(s:str):
    e, d = 0, len(s) - 1
    la = ["a", "e", "i", "o", "u"]
    ls = list(s)

    while e < d:

        while ls[e] not in la:
            e += 1
        while ls[d] not in la:
            d -= 1

        if e < d:
            ls[e], ls[d] = ls[d], ls[e]
            e += 1
            d -= 1
    return "".join(ls)

inverte_vogais("macaco")
```
