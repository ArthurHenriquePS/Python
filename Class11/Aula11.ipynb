```python
# EX. 1
# Pega nome e três notas de uma aluno via input(), guarde em um dicionário, calcule a média e use if
# para imprimir "Aprovado" e "Reprovado"

# Valores de entrada
d = {}
n = input("Digite o nome do aluno: ")

d["nome"] = n
tno = 0

# Loop
for i in range(0, 3):
    # Valor de Entrada
    no = int(input(f"Digite a {i+1}° nota: "))
    d[f"nota{i+1}"] = no
    tno += no

print(tno/(len(d) - 1))
# Condicional
if tno/(len(d)-1) > 7.0:
    # Estrutura de Saída
    print("Aprovado")
else:
    # Estrutura de Saída
    print("Reprovado")


```

```python
# EX. 2
# Crie um dicionário conta com titular e saldo. Peça um valor de saque
# se maior que o saldo, imprima "Saldo insuficiente"

# Estrutura de Entrada
d = {
    "sitular":"Anna",
    "saldo":5000
}
r = int(input("Informe o valor de retirada do saldo: "))

#Condicional
if r > d["saldo"]:
    # Estrututa de Saída
    print("Saldo Insuficiente")
else: 
    # Estrutura de Saída
    print(f"Valor retirado: {r}, saldo restante: {d["saldo"] - r}")
```

```python
# EX. 1
# Crie um algoritmo utilizando dicionário a qual possui um menu com

# 1 - cadastrar
# 2 - consultar
# 3 - sair

# opcao = escolha opcao 1, 2 ou 3

# enquanto opcao nao esta em menu
#     opcao = opcao invalida

# voce escolheu a opcao x do menu

d = {
    1: "cadastrar",
    2: "consultar",
    3: "sair"
}
op = int(input("Digite a opção desejada: "))

while op not in d:
    op = int(input("Opção Inválida, digite novamente: "))

print(f"Opção escolhida {d[op]}")
```
