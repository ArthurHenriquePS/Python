```python
def lista():
    return list(range(1, 100, 1))

def dois_ponteiros(a, l):
    e, d, lf = 0, len(l) - 1, []

    sa = l[e] + l[d]

    while e < d:


        if sa == a:
            lf.append([l[e], l[d]])
            e += 1
            d -= 1
        if sa > a:
            e += 1
        if sa < a:
            d -= 1

    return lf if lf else None


%time dois_ponteiros(100, lista())

```

```python
import pandas as pd

class Estoque:

    def __init__(self, p, e):
        self.p = p
        self.e = e

    def mostrar(self):
        return f"Peso: {self.p} \nEstoque: {self.e}"


p1 = Estoque(300, 10)
p2 = Estoque(400, 10)
p3 = Estoque(400, 20)

lp = [p1, p2, p3]

dados = [
    {
        "preco":n.p,
        "estoque":n.e
    } for n in lp
]

df = pd.DataFrame(dados)
print(df)
```
